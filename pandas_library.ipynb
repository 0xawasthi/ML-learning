## Series in Pandas

It is a one-dimensional array holding data of any type.

```python
import pandas as pd

a = [1, 2, 3, 4]
myvar = pd.Series(a)

print(myvar)
```

**Labels**

If nothing else is specified, the values are labeled with their index number.

In above example- 0, 1, 2, 3 are labels.

```python
#@ Create your own label using index

import pandas as pd

a = [2, 3, 5]
myvar = pd.Series(a, index = ["x", "y", "z"])
print(myvar)
```

```python
print(myvar["x"])
```

```python
#@ Key/Value objects as Series
# Key becomes the label

import pandas as pd

marks = {"sub1": 55, "sub2": 65, "sub3": 75}
myvar = pd.Series(marks)
print(myvar)
```

## Dataframes in Pandas

A Pandas DataFrame is a 2 dimensional data structure, like a 2 dimensional array, or a table with rows and columns.

```python
import pandas as pd

data = {
    "cars": ['BMW', 'Honda', 'Suzuki', 'BYD'],
    "models": ['2007', '2008', '2008', '2022'],
    "passengers": [5, 6, 4, 8]
}
print(data)
```

```python
#@ Convert this dictionary to DataFrame

df = pd.DataFrame(data)
print(df)
```

```python
df_to_csv = df.to_csv("cars.csv")
```

```python
df = pd.read_csv("cars.csv")
df
```

```python
# Locate rows using loc function

print(df.loc[1])
```

```python
# Return 0 and 1

print(df.loc[[0, 1]])
```

```python
# indexing DataFrames

import pandas as pd

data = {
    "calories": [420, 380, 390],
    "duration": [50, 40, 45]
}
df = pd.DataFrame(data, index = ["day1", "day2", "day3"])
print(df)
```

```python
#@ Which was the most ordered items
# group by

i = df.groupby('item_name')
i = i.sum()
i = i.sort_values(['quantity'], ascending=False)
i.head(1)
```

```python
import pandas as pd
df = pd.DataFrame({
    "team":   ["A", "A", "B", "B", "C", "C", "A"],
    "player": ["p1", "p2", "p3", "p4", "p5", "p6", "p7"],
    "points": [10, 15, 8, 12, 20, 5, 7],
    "assists":[3, 5, 2, 4, 6, 1, 2],
})
df
```

```python
df.groupby("team")
```

```python
df.groupby('team')['points'].sum()
```

```python
df.groupby('team')['points'].mean()
```

```python
df.groupby('team')['points'].agg(['mean', 'sum', 'count', 'max', 'min'])
```

```python
df.groupby('team').size()
```
