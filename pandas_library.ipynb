## Series in Pandas

It is a one-dimensional array holding data of any type.

```python
import pandas as pd

a = [1, 2, 3, 4]
myvar = pd.Series(a)

print(myvar)
```

**Labels**

If nothing else is specified, the values are labeled with their index number.

In above example- 0, 1, 2, 3 are labels.

```python
#@ Create your own label using index

import pandas as pd

a = [2, 3, 5]
myvar = pd.Series(a, index = ["x", "y", "z"])
print(myvar)
```

```python
print(myvar["x"])
```

```python
#@ Key/Value objects as Series
# Key becomes the label

import pandas as pd

marks = {"sub1": 55, "sub2": 65, "sub3": 75}
myvar = pd.Series(marks)
print(myvar)
```

```python
#@  EXAMPLE 1

import pandas as pd
pd.set_option('display.max_rows', None)
#pd.set_option('display.max_columns', None)
student_data = pd.DataFrame({
    'school_code': ['s001','s002','s003','s001','s002','s004'],
    'class': ['V', 'V', 'VI', 'VI', 'V', 'VI'],
    'name': ['Alberto Franco','Gino Mcneill','Ryan Parkes', 'Eesha Hinton', 'Gino Mcneill', 'David Parkes'],
    'date_Of_Birth ': ['15/05/2002','17/05/2002','16/02/1999','25/09/1998','11/05/2002','15/09/1997'],
    'age': [12, 12, 13, 13, 14, 12],
    'height': [173, 192, 186, 167, 151, 159],
    'weight': [35, 32, 33, 30, 31, 32],
    'address': ['street1', 'street2', 'street3', 'street1', 'street2', 'street4']},
    index=['S1', 'S2', 'S3', 'S4', 'S5', 'S6'])

print("Original DataFrame:")
print(student_data)
print('\nSplit the said data on school_code wise:')
result = student_data.groupby(['school_code'])
for name,group in result:
    print("\nGroup:")
    print(name)
    print(group)
print("\nType of the object:")
print(type(result))
```

```python
#Write a Pandas program to group the dataframe by school code and then iterate over each group to print the first row.

result = student_data.groupby(['school_code'])
for name,group in result:
    print("\nGroup:")
    print(name)
    print(group.head(1))
```

```python
# Write a Pandas program to group the dataframe by school code and then extract the list of unique class values for each school.

unique = student_data.groupby('school_code')['class'].unique()
print(unique)
```

```python
# Write a Pandas program to group the dataframe by school code and check if the GroupBy object is an instance of pandas.core.groupby.generic.DataFrameGroupBy.

result = student_data.groupby(['school_code'])
print(isinstance(result, pd.core.groupby.generic.DataFrameGroupBy))
```

```python
# Write a Pandas program to group the dataframe by school code and then count the number of students in each group.

result = student_data.groupby(['school_code']).size()
print(result)
```

```python
#@ EXAMPLE 2

import pandas as pd
pd.set_option('display.max_rows', None)
#pd.set_option('display.max_columns', None)
student_data = pd.DataFrame({
    'school_code': ['s001','s002','s003','s001','s002','s004'],
    'class': ['V', 'V', 'VI', 'VI', 'V', 'VI'],
    'name': ['Alberto Franco','Gino Mcneill','Ryan Parkes', 'Eesha Hinton', 'Gino Mcneill', 'David Parkes'],
    'date_Of_Birth ': ['15/05/2002','17/05/2002','16/02/1999','25/09/1998','11/05/2002','15/09/1997'],
    'age': [12, 12, 13, 13, 14, 12],
    'height': [173, 192, 186, 167, 151, 159],
    'weight': [35, 32, 33, 30, 31, 32],
    'address': ['street1', 'street2', 'street3', 'street1', 'street2', 'street4']},
    index=['S1', 'S2', 'S3', 'S4', 'S5', 'S6'])

print("Original DataFrame:")
print(student_data)
print('\nMean, min, and max value of age for each value of the school:')
grouped_single = student_data.groupby('school_code').agg({'age': ['mean', 'min', 'max']})
print(grouped_single)
```

```python
# Write a Pandas program to group the dataframe by school code and then determine the minimum and maximum age values per group.

k = student_data.groupby('school_code')['age'].agg(['min', 'max'])
print(k)
```

```python
# Write a Pandas program to group the dataframe by school code and apply multiple aggregations (mean, min, max) on the age column.

k = student_data.groupby('school_code')['age'].agg(['mean', 'min', 'max'])
print(k)
```

## Exploratory Data Analysis (EDA)

### 1. Data Type Conversion and Cleaning

```python
# Clean and convert 'actualgross', 'adjusted_gross', and 'average_gross' to numeric
df['actualgross'] = df['actualgross'].astype(str).str.replace('NPR', '', regex=False).str.replace(',', '', regex=False).astype(float)
df['adjusted_gross'] = df['adjusted_gross'].astype(str).str.replace('NPR', '', regex=False).str.replace(',', '', regex=False).astype(float)
df['average_gross'] = df['average_gross'].astype(str).str.replace('NPR', '', regex=False).str.replace(',', '', regex=False).astype(float)

# Handle 'peak' and 'all_time_peak' columns
df['peak'] = df['peak'].replace('delete_me', np.nan).astype(float)
df['all_time_peak'] = df['all_time_peak'].replace('do_not_delete_me', np.nan).astype(float)

df.info()
```

### 3. Visualizations

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Distribution of 'actualgross'
plt.figure(figsize=(10, 6))
sns.histplot(df['actualgross'], kde=True)
plt.title('Distribution of Actual Gross Revenue')
plt.xlabel('Actual Gross Revenue')
plt.ylabel('Frequency')
plt.show()

# Distribution of 'shows'
plt.figure(figsize=(10, 6))
sns.histplot(df['shows'], kde=True)
plt.title('Distribution of Number of Shows')
plt.xlabel('Number of Shows')
plt.ylabel('Frequency')
plt.show()
```

```python
# Relationship between 'artist' and 'actualgross'
plt.figure(figsize=(14, 7))
sns.barplot(x='artist', y='actualgross', data=df.sort_values('actualgross', ascending=False))
plt.title('Actual Gross Revenue by Artist')
plt.xlabel('Artist')
plt.ylabel('Actual Gross Revenue')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
